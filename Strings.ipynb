```python
# Length of a string without len()
s = input("enter the string")

counter = 0

for i in s:
    counter += 1

print("length of string is", counter)
```

```python
# Extract username from email
s = input("enter the email")

pos = s.index("@")
print(s[0:pos])
```

```python
# Count character frequency
s = input("enter the email")
term = input("what would like to search for")

counter = 0
for i in s:
    if i == term:
        counter += 1

print("frequency", counter)
```

```python
# Remove character from string
s = input("enter the string")
term = input("what would like to remove")

result = ""

for i in s:
    if i != term:
        result = result + i

print(result)
```

```python
# Palindrome check
s = input("enter the string")
flag = True
for i in range(0, len(s) // 2):
    if s[i] != s[len(s) - i - 1]:
        flag = False
        print("Not a Palindrome")
        break

if flag:
    print("Palindrome")
```

```python
# Count/extract words without split()
s = input("enter the string")
L = []
temp = ""
for i in s:
    if i != " ":
        temp = temp + i
    else:
        L.append(temp)
        temp = ""

L.append(temp)
print(L)
```

```python
# Title case without title()
s = input("enter the string")

L = []
for i in s.split():
    L.append(i[0].upper() + i[1:].lower())
print(L)
print(" ".join(L))
```

```python
# Integer to string conversion
number = int(input("enter the number"))

digits = "0123456789"
result = ""
while number != 0:
    result = digits[number % 10] + result
    number = number // 10

print(result)
print(type(result))
```
