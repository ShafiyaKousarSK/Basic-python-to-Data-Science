**Sequence Sum**
1/1! + 2/2! + 3/3! + ...+n/n!

```python
n=int(input('enter n'))
sum=0
factorial=1
for i in range(1,n+1):
  factorial=i*factorial
  sum+=i/factorial
print(sum)
```
