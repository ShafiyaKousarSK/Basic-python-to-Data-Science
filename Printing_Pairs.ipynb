**Pairs **
1) All Possible Pairs from 1 to n

```python
n=int(input('enter number'))
for i in range(1,n+1):
  for j in range(1,n+1):
    print(i,j)
```

2. Unique Combination without duplicate values

```python
n=int(input('enter number'))
for i in range(1,n+1):
  for j in range(i+1,n+1):
    print(i,j)
```
