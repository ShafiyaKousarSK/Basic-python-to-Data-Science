1. print stars

```python
r=int(input('enter number of rows '))
for i in range(1,r+1):
  for j in range(1,i+1):
    print("*",end=" ")
  print()
```

2. print numbers

```python
r=int(input('enter number of rows '))
for i in range(1,r+1):
  for j in range(1,i+1):
    print(j,end=" ")
  print()
```

3. Palindrome number Pyramid

```python
r=int(input('enter number of rows '))
for i in range(1,r+1):
  for j in range(1,i+1):
    print(j,end="")
  for k in range(i-1,0,-1):
    print(k,end="")
  print()
```
